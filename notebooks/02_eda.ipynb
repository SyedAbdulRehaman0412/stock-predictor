```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/tcs.csv", index_col="Date", parse_dates=True)
df.head()
```

```python
df["SMA_20"] = df["Close"].rolling(20).mean()
df["SMA_50"] = df["Close"].rolling(50).mean()

plt.figure(figsize=(12, 5))
plt.plot(df["Close"], label="Close")
plt.plot(df["SMA_20"], label="20-day average")
plt.plot(df["SMA_50"], label="50-day average")
plt.legend()
plt.title("TCS price with moving averages")
plt.show()
```

```python
df["Return"] = df["Close"].pct_change()

plt.figure(figsize=(12, 4))
plt.plot(df["Return"])
plt.title("Daily returns")
plt.show()

df["Return"].describe()
```

```python
df["Return"].hist(bins=60, figsize=(8, 4))
plt.title("Distribution of daily returns")
plt.show()
```

```python
plt.figure(figsize=(12, 3))
plt.plot(df["Volume"])
plt.title("Trading volume")
plt.show()
```
