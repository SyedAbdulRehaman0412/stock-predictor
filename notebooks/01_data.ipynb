```python
import yfinance as yf

df = yf.download("TCS.NS", start="2020-01-01")
print(df.tail())
```

```python
print(df.shape)
df.info()
```

```python
import pandas as pd

df.columns = df.columns.get_level_values(0)
df = df.dropna()
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.plot(df["Close"])
plt.title("TCS Closing Price")
plt.xlabel("Date")
plt.ylabel("Price (INR)")
plt.show()
```

```python
df.to_csv("../data/tcs.csv")
```
